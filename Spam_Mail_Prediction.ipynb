# Spam Mail Prediction using Machine Learning

**Project type:** Binary text classification using TF-IDF and Logistic Regression.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

## 1. Data Collection and Inspection

```python
raw_mail_data = pd.read_csv("mail_data.csv")
print("Dataset shape:", raw_mail_data.shape)
print("Class distribution:")
print(raw_mail_data["Category"].value_counts())
display(raw_mail_data.head())
```

## 2. Label Encoding and Cleaning

```python
mail_data = raw_mail_data.where(pd.notnull(raw_mail_data), "")
mail_data["Category"] = mail_data["Category"].map({"spam":0, "ham":1})
X = mail_data["Message"]
Y = mail_data["Category"].astype(int)
print("Encoded labels: spam=0, ham=1")
```

## 3. Train/Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3, stratify=Y)
print("Training samples:", X_train.shape[0])
print("Test samples:", X_test.shape[0])
```

## 4. TF-IDF Feature Extraction

```python
feature_extraction = TfidfVectorizer(min_df=1, stop_words="english", lowercase=True)
X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)
print("Training feature matrix:", X_train_features.shape)
print("Test feature matrix:", X_test_features.shape)
```

## 5. Train Logistic Regression

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_features, Y_train)
print("Model training completed.")
```

## 6. Evaluation

```python
prediction_on_training_data = model.predict(X_train_features)
prediction_on_test_data = model.predict(X_test_features)
train_accuracy = accuracy_score(Y_train, prediction_on_training_data)
test_accuracy = accuracy_score(Y_test, prediction_on_test_data)
precision = precision_score(Y_test, prediction_on_test_data)
recall = recall_score(Y_test, prediction_on_test_data)
f1 = f1_score(Y_test, prediction_on_test_data)
print(f"Training accuracy: {train_accuracy:.4f}")
print(f"Test accuracy: {test_accuracy:.4f}")
print(f"Test precision (ham=1): {precision:.4f}")
print(f"Test recall (ham=1): {recall:.4f}")
print(f"Test F1 (ham=1): {f1:.4f}")
print("Confusion matrix:")
print(confusion_matrix(Y_test,prediction_on_test_data))
print("Classification report:")
print(classification_report(Y_test,prediction_on_test_data,target_names=["spam","ham"]))
```

## 7. Predictive System

```python
def predict_mail(message):
    features = feature_extraction.transform([message])
    pred = int(model.predict(features)[0])
    return "Ham mail" if pred == 1 else "Spam mail"

sample = "Congratulations! You have won a free prize. Call now!"
print("Sample:", sample)
print("Prediction:", predict_mail(sample))
```

## 8. Conclusion
The completed model converts email text to TF-IDF features and uses Logistic Regression to classify messages as spam or ham. The test metrics provide a measurable baseline for the supplied dataset.
